Prompt:

create two graphs that both look at the relationship between transations with pins and fradualent transactions have the first graph be a histogram with red and green bars and the second graph be a stacked bar graph

for the first plot have the x be if the used pin number and the hue be fraudualnt or not 

make it a countplot

```python
plt.figure(figsize=(7,4))
sns.countplot(
    data=df_cleaned,
    x='used_pin_number',
    hue='fraud',
    palette={0: 'green', 1: 'red'}
)

plt.title('Transactions Using PIN vs Fraudulent Transactions', fontsize=14)
plt.xlabel('Used PIN Number', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend(title='Fraud (0 = No, 1 = Yes)')
plt.show()
```

```python
pin_fraud_counts = pd.crosstab(df_cleaned['used_pin_number'], df_cleaned['fraud'])

# Plot stacked bar chart
pin_fraud_counts.plot(
    kind='bar',
    stacked=True,
    color=['green', 'red'],
    figsize=(7,4)
)
plt.title('Transactions Using PIN vs Fraudulent Transactions (Stacked Bar)', fontsize=14)
plt.xlabel('Used PIN Number', fontsize=12)
plt.ylabel('Count', fontsize=12)
plt.legend(title='Fraud', labels=['No', 'Yes'])
plt.show()
```

Prompt: with a data set name df_cleaned create a count plot that sets the hue to the fraud column and x to be used_chip

AI Used: Chat GPT 5-mini

```python
percent_df = df_cleaned.groupby(['used_chip', 'fraud']).size().reset_index(name='count')
total_per_chip = percent_df.groupby('used_chip')['count'].transform('sum')
percent_df['percent'] = percent_df['count'] / total_per_chip * 100

# Plot
sns.barplot(data=percent_df, x='used_chip', y='percent', hue='fraud')

plt.ylabel('Percentage (%)')
plt.xlabel('Used Chip')
plt.title('Percentage of Fraud by Used Chip')
plt.show()

counts = df_cleaned.groupby(['used_chip', 'fraud']).size().unstack(fill_value=0)

# Convert to percentages
percentages = counts.div(counts.sum(axis=1), axis=0) * 100

# Plot stacked bar chart
ax = percentages.plot(kind='bar', stacked=True, color=['green', 'red'])

plt.ylabel('Percentage (%)')
plt.xlabel('Used Chip')
plt.title('Stacked Percentage of Fraud by Used Chip')
plt.legend(title='Fraud')
plt.show()
```
